```python
from langchain.agents import create_agent
from langchain_ollama import ChatOllama
from langchain_community.tools.tavily_search import TavilySearchResults
```

```python
from dotenv import load_dotenv
load_dotenv()
import os
```

```python
llm=ChatOllama(model="gpt-oss:120b-cloud")
```

```python
tavily_tool=TavilySearchResults(max_results=3)
```

```python
my_agent= my_agent=create_agent(
    model=llm,
    tools=[tavily_tool],
    system_prompt="You are a helpful assistant"

)
```

```python
my_agent
```

```python
response=my_agent.invoke({"messages":"who won ind vs eng yesterday(july 14 2026) ODI cricket match"})
```

```python
response 
```

```python
from langchain_tavily import TavilySearch
tavily_search_tool=TavilySearch(
    max_results=3,
    topic="general",
    )
agent=create_agent(model=llm, tools=[tavily_search_tool])
user_input="what nation hosted the Euro 2024 include Only Wikipedia results"
response=agent.invoke({"messages":user_input})
response
```
