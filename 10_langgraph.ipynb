Langgraph using llm groq model 

```python
from langgraph.graph import StateGraph, START, END
from typing import TypedDict
from langchain_groq import ChatGroq
from dotenv import load_dotenv 
load_dotenv() 
```

```python
llm= ChatGroq(model="openai/gpt-oss-120b")
```

```python
class StateClass(TypedDict):
    question: str
    answer: str 
```

```python
def call_llm(state: StateClass) -> StateClass: 
    state['answer'] =llm.invoke(state['question']).content 
    return state
```

```python
graph=StateGraph(StateClass)
graph.add_node("my_node", call_llm)

graph.add_edge(START, "my_node")
graph.add_edge( "my_node",END)
your_graph=graph.compile()
your_graph 

```

```python
result=your_graph.invoke({"question": "What is langgraph in one sentence"}) 
```
