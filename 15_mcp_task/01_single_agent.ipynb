```python
from langchain_ollama import ChatOllama
from langchain_core.messages import HumanMessage
from langgraph.graph import StateGraph, MessagesState, START, END
from typing import TypedDict 
```

```python
llm = ChatOllama(model = "gpt-oss:120b-cloud")
```

```python
class StateClass(TypedDict):
    topic: str
    information :str
    answer : str
#this is stateClass created by us to define the state of graph we are not using this 
```

```python
print(MessagesState.__annotations__)
#we are using prebuilt LangGraph’s built-in MessagesState.
```

```python
#creating the agent node

def agent_node(state:MessagesState):
    """
    Receives the current graph state,
    sends the messages to the LLM,
    and returns the AI response.
    """

    response = llm.invoke(state["messages"])

    return {
        "messages":[response]
    }
```

```python
graph_builder = StateGraph(MessagesState)
```

```python
graph_builder.add_node("agent",agent_node)
graph_builder.add_edge(START,"agent")
graph_builder.add_edge("agent",END)
```

```python
single_agent = graph_builder.compile()
print(single_agent)
```

```python
result = single_agent.invoke({
    "messages":[{"role":"user","content":"What is human in the loop? explain in short "}]
})
```
