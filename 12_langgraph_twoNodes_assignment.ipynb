```python
from langgraph.graph import StateGraph, START, END
from typing import TypedDict
from langchain_groq import ChatGroq
from dotenv import load_dotenv 
load_dotenv()  
```

```python
llm= ChatGroq(model="openai/gpt-oss-120b")
```

```python
class StateClass(TypedDict):
    topic: str
    research: str
    blog: str
```

```python
def blog(state: StateClass)->StateClass:
    question = f"Generate attractive blog on given research :{state['research']}"
    answer = llm.invoke(question)
    state['blog'] = answer.content
    return state
```

```python
def research(state: StateClass)->StateClass:
    prompt = f"Do Research on topic: {state['topic']}"
    answer = llm.invoke(prompt)
    state['research'] = answer.content
    return state
```

```python
graph = StateGraph(StateClass)
graph.add_node("research", research)
graph.add_node("blog", blog)
graph.add_edge(START, "research")
graph.add_edge("research", "blog")
graph.add_edge("blog", END)
graph = graph.compile()
```

```python
res = graph.invoke({"topic": "Artificial Intelligence in Healthcare"})
res
```
